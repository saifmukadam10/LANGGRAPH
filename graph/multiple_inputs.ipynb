### Multiple Input Agent

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
import math
```

```python
class AgentState (TypedDict):
    values : List[int]
    name : str
    operation: str
    results : str

```

```python
def arithmatic (state : AgentState) -> AgentState:
    """Result Operation"""
    if state['operation'] == "*":
        state['results'] = f"Hi {state['name']}, your score is { math.prod(state['values'])}"
    else:
        state['results'] = f"Hi {state['name']}, your score is { sum(state['values'])}"

    return state
        
```

```python
graph = StateGraph(AgentState)

graph.add_node("Answer", arithmatic)
graph.set_entry_point("Answer")
graph.set_finish_point("Answer")

app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
ans = app.invoke({'name': "Saif", 'values': [1,2,3,4], 'operation': "+"})
ans["results"]
```
